```python
def combine_parquet(folder_name):
    all_days=[]
    for date in sorted(os.listdir(main_path)):
        folder_path = os.path.join(main_path,date,"data","raw",folder_name)
        if os.path.exists(folder_path):
            print("Processing:", date,folder_name)
            daily_files =[]
            for file in os.listdir(folder_path):
                if file.endswith(".parquet"):
                    file_path = os.path.join(folder_path,file)
                    df = pd.read_parquet(file_path)
                    daily_files.append(df)
            #combine all files of one day
            if len(daily_files)>0:
                day_df = pd.concat(daily_files,ignore_index=True)
                day_df["date"] = date
                all_days.append(day_df)
                print("Day shape:", day_df.shape)
    #combine all days
    final_df = pd.concat(all_days,ignore_index=True)
    return final_df
```

```python
def combine_match_section(section_name):
        
        all_days = []

        for date in sorted(os.listdir(main_path)):
            folder_path = os.path.join(main_path,date,"data","raw","raw_match_parquet")
            if os.path.exists(folder_path):
                daily_files = []
                for file in os.listdir(folder_path):
                    if not file.endswith(".parquet"):
                        continue
                    file_name =file.removesuffix(".parquet")
                    prefix = section_name + "_"
                    if file_name.startswith(prefix):
                        remaining_part=file_name[len(prefix):]
                        if remaining_part.isdigit():
                            file_path = os.path.join(folder_path,file)
                            df = pd.read_parquet(file_path)
                            daily_files.append(df)

                #combine this section for this day 
                if len(daily_files)>0:
                    day_df =pd.concat(daily_files,ignore_index=True)
                    day_df["date"]=date
                    all_days.append(day_df)
                    print(date,section_name,day_df.shape)
        #combine all days 
        if len(all_days)>0:
            final_df=pd.concat(all_days,ignore_index=True)
            return final_df
        else:
            print("No files found for:",section_name)
            return pd.DataFrame()
```

```python
match_event_info = combine_match_section("event")
match_event_info.to_parquet(
    "data/processed/match_event_info.parquet",
    index=False
)
```
