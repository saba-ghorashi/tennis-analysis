### 2. Missing Values Analysis

```python
for name,df in datasets.items():
    print("\n",name)
    print(df.isna().sum().sort_values(ascending=False))
```

```python
print("MatchHomeScoreInfoMatchHomeScoreInfo shape:", MatchHomeScoreInfo.shape)
print("MatchAwayScoreInfo shape:", MatchAwayScoreInfo.shape)


# Columns with missing values to investigate
score_columns = [
    "period_4",
    "period_5",
    "period_4_tie_break",
    "period_5_tie_break",
    "normal_time"
]


# Missing value summary for MatchHomeScoreInfo
print("\nMissing values in MatchHomeScoreInfo:")
home_missing = pd.DataFrame({
    "Missing Count": MatchHomeScoreInfo[score_columns].isnull().sum(),
    "Missing Percentage": MatchHomeScoreInfo[score_columns].isnull().mean() * 100
})

print(home_missing)


# Missing value summary for MatchAwayScoreInfo
print("\nMissing values in MatchAwayScoreInfo:")
away_missing = pd.DataFrame({
    "Missing Count": MatchAwayScoreInfo[score_columns].isnull().sum(),
    "Missing Percentage": MatchAwayScoreInfo[score_columns].isnull().mean() * 100
})

print(away_missing)
```

## Missing Values in Match Score Information

The `MatchHomeScoreInfo` and `MatchAwayScoreInfo` DataFrames contain 35,053 rows and 15 columns.

The missing value analysis showed that the following columns contain missing values in all records (100% missing):

- `period_4`
- `period_5`
- `period_4_tie_break`
- `period_5_tie_break`
- `normal_time`

These columns do not provide any information in the current dataset because all values are missing. Therefore, they can be removed during the data cleaning process.

The missing values in `period_4` and `period_5` are expected because most tennis matches follow a best-of-three sets format, meaning that fourth and fifth sets usually do not occur.

```python
statistics_missing = statistics[
    statistics["home_total"].isna() |
    statistics["away_total"].isna()
]

statistics_missing.head()
```

```python
statistics_complete = statistics[
    statistics["home_total"].notna() |
    statistics["away_total"].notna()
]

statistics_complete.head()
```
