## Missing Values

After checking the missing values in all DataFrames, the following observations were found:

- The `match` DataFrame contains a large number of missing values across several columns.
- In the `odds` DataFrame, the `winning` column contains missing values.
- In the `statistics` DataFrame, the `home_total` and `away_total` columns contain missing values.

These missing values need to be investigated before deciding whether to remove them, replace them, or keep them.

```python
for name,df in datasets.items():
    print(name,df.duplicated().sum())
```

Duplicate records should be investigated because they may distort the analysis. However, they should not be removed automatically before understanding what one row represents in each DataFrame.

### Investigation of Duplicate Match IDs

The `match` DataFrame contains 316,802 rows but only 16,873 unique `match_id` values. This indicates that multiple rows can belong to the same tennis match.

Therefore, repeated `match_id` values should not automatically be considered duplicate records.

There are also 5,870 exact duplicate rows, where all column values are identical. These records require further investigation before removal.

The next step is to examine multiple rows belonging to the same `match_id` to understand what each row represents.

```python
summary=[]
for col in match.columns:
    summary.append([col,match[col].dtypes,match[col].nunique(),match[col].isna().sum(),round(match[col].isna().mean()*100.2)])
summary =pd.DataFrame(summary,columns=["columns","dtype","unique_values","missing","missing_percent"])
summary
```
